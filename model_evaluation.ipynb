```python
import json
import pandas as pd
import numpy as np
from sklearn.metrics import f1_score, jaccard_score
```

```python
with open("final_dataset_g2.0.json", "r") as f:
    data1 = json.load(f)

with open("final_dataset_g1.5.json", "r") as f:
    data2 = json.load(f)
```

```python
data1
```

```python
combined = []
for idx, (entry1, entry2) in enumerate(zip(data1, data2), start=1):
    combined.append({
        "incident_id": idx,
        "incident_summary":entry1["incident_summary"],
        "category_flow": entry1["category_flow"],  # Ground truth
        "llm_suggestion_1.5": entry2["llm_suggestion"],
        "llm_suggestion_2.0": entry1["llm_suggestion"]
    })

# Convert to DataFrame
df = pd.DataFrame(combined)
```

```python
print(df)
```

```python
with open('incident.txt', 'r') as f:
    incidents = [line.strip().strip('"') for line in f.readlines()]
```

```python
df.insert(1, 'incident_name', incidents)  # Insert after incident_id

# Save back to CSV
df.to_csv('attack_incident_updated.csv', index=False)
```

```python
all_labels = ["AAT", "BPA", "DCA", "EFA", "NDA", "OIA", "PSI", "SPI"]
label_to_index = {label: idx for idx, label in enumerate(all_labels)}
```

```python
all_labels
```

```python
label_to_index
```

```python
def binarize(labels):
    vec = [0] * len(all_labels)
    for label in labels:
        if label in label_to_index:
            vec[label_to_index[label]] = 1
    return vec

# Prepare binary matrices
y_true1 = np.array([binarize(entry["category_flow"]) for entry in data1])
y_pred1 = np.array([binarize(entry["llm_suggestion"]) for entry in data1])
```

```python
y_true2 = np.array([binarize(entry["category_flow"]) for entry in data2])
y_pred2 = np.array([binarize(entry["llm_suggestion"]) for entry in data2])
```

```python
y_true1
```

```python
y_pred1
```

```python
micro1_f1 = f1_score(y_true1, y_pred1, average='micro')
macro1_f1 = f1_score(y_true1, y_pred1, average='macro')
jaccard1_micro = jaccard_score(y_true1, y_pred1, average='micro')
jaccard1_macro = jaccard_score(y_true1, y_pred1, average='macro')



micro2_f1 = f1_score(y_true2, y_pred2, average='micro')
macro2_f1 = f1_score(y_true2, y_pred2, average='macro')
jaccard2_micro = jaccard_score(y_true2, y_pred2, average='micro')
jaccard2_macro = jaccard_score(y_true2, y_pred2, average='macro')
```

```python
print("=== Evaluation Results for 2.0 ===")
print(f"Micro F1 Score     : {micro1_f1:.4f}")
print(f"Macro F1 Score     : {macro1_f1:.4f}")
print(f"Jaccard (Micro)    : {jaccard1_micro:.4f}")
print(f"Jaccard (Macro)    : {jaccard1_macro:.4f}")


print("=== Evaluation Results for 1.5===")
print(f"Micro F1 Score     : {micro2_f1:.4f}")
print(f"Macro F1 Score     : {macro2_f1:.4f}")
print(f"Jaccard (Micro)    : {jaccard2_micro:.4f}")
print(f"Jaccard (Macro)    : {jaccard2_macro:.4f}")
```
